```python
def tabla_de_multiplicar(numero):
  resultado = []
  for i in range(11):
    resultado.append(i*numero)
  return resultado

res = tabla_de_multiplicar(5)
print(res)
```

```python
def calcular(n=1):
  tabla = []
  for i in range(0, 11):
    tabla.append(f'{n} x {i} = {n * i}')
  return tabla

def calcular_todas():
  for i in range(0, 11):
    print(f'tabla del {i}:')
    tabla = (calcular(i))
    for j in tabla:
      print(j)
    print('-' * 10)

calcular_todas()
```

```python
def cuadrado(x):
  return x**2

def cubo(x):
  return x**3

enteros = [1,2,3,4,5]
funciones =[cuadrado,cubo]

for e in enteros:
  valores = list(map(lambda x: x(e), funciones))
  print(valores)
```
